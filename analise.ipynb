```python
import pandas as pd


# Para carregar o dado direto no pandas
df = pd.read_csv('brasileirao_serie_a.csv')


# Gols como mandante no ano de 2021 do time Palmeiras
ano_2021_pal_mand = df[(df['ano_campeonato'] == 2021) & (df['time_mandante'] == 'Palmeiras')]['gols_mandante'].sum()
# Gols como visitante no ano de 2021 do time Palmeiras
ano_2021_pal_visit = df[(df['ano_campeonato'] == 2021) & (df['time_visitante'] == 'Palmeiras')]['gols_visitante'].sum()
# Defesas do palmeiras como mandante
ano_2021_pal_def_mand = df[(df['ano_campeonato'] == 2021) & (df['time_mandante'] == 'Palmeiras')]['defesas_mandante'].sum()
# Defesas do palmeiras como visitante
ano_2021_pal_def_visit = df[(df['ano_campeonato'] == 2021) & (df['time_visitante'] == 'Palmeiras')]['defesas_mandante'].sum()
# total de gols na temporada 2021 do palmeiras
totGolsPal21 = ano_2021_pal_mand + ano_2021_pal_visit
# total de defesas na temporada 2021 do palmeiras
defsPal21 = ano_2021_pal_def_mand + ano_2021_pal_def_visit

print("\n============= Estatísticas Temporada 2021 Palmeiras ===================")
print(f'Total de gols marcados: {totGolsPal21} gols')
# 38 é o número de jogos totais no campeonato brasileiro
print(f'Média de gols por jogos: {totGolsPal21 / 38:.2f} gols')

print(f'Total de defesas realizadas: {defsPal21} defesas')

print(f'Média de defesas realizadas por jogo: {defsPal21 / 38:.2f} defesas')
```
